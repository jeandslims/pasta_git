```python
import pandas as pd
```

```python
abas = pd.read_excel(r"agregado_resultado_planilha_base_2016_2025_BR_CE_FORTALEZA_BA_NE_SP_MG (1).xlsx",
                         sheet_name=None)
```

```python
def trata(df):
    df_filtrado = df[
        (df["Fonte"].str.contains("PNADC_01")) &
         (df["Local"] == "Brasil") &
         (df["Sexo"].isna()) &
         (df["Faixa etária"].isna()) &
         (df["Nível de instrução"].isna()) &
         (df["Raça"].isna())
         ]
    return df_filtrado
```

```python
abas_tratadas = {nome: trata(df) for nome, df in abas.items() if nome != "Sheet 1"}
```

```python
df_final = pd.concat(
    [df.assign(aba=nome) for nome, df in abas_tratadas.items()],
    ignore_index=True
)
```

```python
df_final["Fonte"] = df_final["Fonte"].str.replace("PNADC_01", "1° Tri. ")
```

```python
df_final.set_index("Fonte", inplace= True)
```

```python
df_final.to_excel("primeiros_tri.xlsx")
```
