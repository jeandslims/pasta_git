```python
import pandas as pd
import plotly.graph_objects as go
```

```python
url_hab_tot = 'https://apisidra.ibge.gov.br/values/t/6472/p/all/v/5933/n1/1/n2/2/n3/23'
df_geral = pd.read_json(url_hab_tot)
```

```python
df_geral.columns = df_geral.iloc[0]
df_geral.drop(0, inplace=True)
```

```python
df_geral["Valor"] = pd.to_numeric(df_geral["Valor"])
```

```python
df_geral = df_geral[['Trimestre', 'Brasil, Grande Região e Unidade da Federação','Valor']]
```

```python
df_geral['Trimestre'] = df_geral['Trimestre'].str.replace("º trimestre ", "/")
```

```python
df_geral = df_geral.pivot(
    index = 'Trimestre',
    columns = 'Brasil, Grande Região e Unidade da Federação',
    values  = 'Valor'
)
```

```python
df_geral = df_geral.reset_index()
```

```python
df_geral['Data'] = pd.to_datetime(df_geral['Trimestre'], format='%m/%Y')
```

```python
df_geral = df_geral.sort_values('Data')
```

```python
df_geral = df_geral.drop(columns=['Data'])
```

```python
df_geral.set_index('Trimestre')
```
