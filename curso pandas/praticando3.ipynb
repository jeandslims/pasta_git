```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
tabela_url = f'https://sidra.ibge.gov.br/tabela/4093#resultado'
browsers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}
page = requests.get(tabela_url, headers = browsers)

soup = BeautifulSoup(page.content, 'html.parser')
tab_completa = soup.find_all("td", class_ = "yf-1m2i7s2")
```
